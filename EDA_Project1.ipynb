```python
#replacing null values which have question mark(?)
data = data.replace(" ?" ,None)
```

```python
print(data.isna().sum())
```

```python
#cleaning
#identify NaN Values
data.isna()
```

```python
#Filling CONSTANT VALUES
data['Workclass'] = data['Workclass'].fillna("Unknown")
print(data)
```

```python
#Filling Mean
data['Age'].mean()
```

```python
#Filling Median
data['Age'].median()
```

```python
#Filling Mode
data['Workclass'].fillna(data['Workclass'].mode()[0], inplace=True)
```

```python
#cleaning NULL Values
#data.dropna()
```

```python
#Delete the row or column
#data.dropna(axis=0)
```

```python
#Backward Filling
data['Age']=data['Age'].bfill()
data['Hours per Week'] =data['Hours per Week'].bfill()
print(data)
```

```python
#Forward Filling
data['Age']=data['Age'].ffill()
data['Hours per Week'] =data['Hours per Week'].ffill()
print(data)
```

```python
#Interpolate:
data['Age']=data['Age'].interpolate()
print(data)
```

```python
#Detecting Outliers bounds(lower&upper)
Q1 = data['Age'].quantile(0.25)
Q3 = data['Age'].quantile(0.75)
IQR = Q3-Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print(lower_bound)
print(upper_bound)
```

```python
#Identify IQR outliers
iqr_outliers = data[
 (data['Age'] < lower_bound) |
 (data['Age'] > upper_bound)
]
print("IQR Outliers:",len(iqr_outliers))
```

```python
import pandas as pd
import numpy as np
from scipy.stats import zscore
from scipy.stats.mstats import winsorize
```

```python
#Z-Score Method
data['Age_zscore'] = zscore(data['Age'])
z_outliers = data[
    data['Age_zscore'].abs()>3
]
print("Z-Score Outliers:",len(z_outliers))
```

```python
#Z-Score Trimming
data_trimmed = data[data['Age_zscore'].abs() <= 3]
print(data_trimmed)
```

```python
#Winsorization
data['Age_winsorized'] = winsorize(
    data['Age'],
    limits=[0,0.3])
print(data.head())
```

```python
#Outliers on Hours per week Column
#1.
Q1 = data['Hours per Week'].quantile(0.25)
Q3 = data['Hours per Week'].quantile(0.75)
IQR = Q3-Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print(lower_bound)
print(upper_bound)
```

```python
#2.
iqr_outliers = data[
 (data['Hours per Week'] < lower_bound) |
 (data['Hours per Week'] > upper_bound)
]
print("IQR Outliers:",len(iqr_outliers))
```

```python
#3.
#Z-Score Method
data['Hours per Week_zscore'] = zscore(data['Hours per Week'])
z_outliers = data[
    data['Hours per Week_zscore'].abs()>3
]
print("Z-Score Outliers:",len(z_outliers))
```

```python
#4.
#Z-Score Trimming
data_trimmed = data[data['Hours per Week_zscore'].abs() <= 3]
print(data_trimmed)
```

```python
#5.Winsorization
data['Hours per Week_winsorized'] = winsorize(
    data['Hours per Week'],
    limits=[0,0.3])
print(data.head())
```
